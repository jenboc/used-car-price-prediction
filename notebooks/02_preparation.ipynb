# 2. Data Preparation

In this notebook we aim to do two things with the raw data `data/all_car_adverts.csv`.

1. Clean the data by fixing the issues we identified in our initial exploration.
2. Transform some of the columns such that they can be used in a machine learning model, i.e. Feature Engineering.


```python
import pandas as pd
import numpy as np
```

```python
df_original = pd.read_csv("../data/all_car_adverts.csv")
df = df_original.copy()
df.head()
```

## 2.1. Data Wrangling
### 2.1.1. Column Naming

We can drop the `Unnamed: 0` column as this is just acting as an index, which we already have.

```python
df.drop("Unnamed: 0", axis=1, inplace=True)
df.head()
```

Then, as personal preference, we change the naming convention from `snake_case` to `PascalCase`. We also change `FeulType` to `FuelType` (fixing the typo) and rename `Ulez` and `ULEZ` as this is an abbreviation.

```python
original_headers = list(df.columns.values)
renaming_dict = {h: h.replace("_", " ").title().replace(" ", "") for h in original_headers}
renaming_dict["feul_type"] = "FuelType"
renaming_dict["ulez"] = "ULEZ"

df.rename(columns=renaming_dict, inplace=True)
df.head()
```

### 2.1.2. Data Types

A lot of the columns have not been assigned the correct data type. We correct these now.

```python
df.dtypes
```

One of the first things we notice is the sheer amount of `int64` columns which actually represent boolean, or yes/no, flags. We convert these to the correct type.

```python
# Check and change boolean columns to the correct type
def check_and_change(col_name):
    col = df[col_name]

    if not pd.api.types.is_integer_dtype(col):
        return False
    
    if col.max() > 1 or col.min() < 0:
        return False
    
    df[col_name] = df[col_name].astype(bool)
    return True


for col_name in df.columns.values:
    if check_and_change(col_name):
        print(f"{col_name} changed to boolean type")

df.head()
```

The other issue is that `NumOwner` is a float, when it should really be an integer.

```python
df["NumOwner"] = df["NumOwner"].astype(pd.Int64Dtype())
df.head()
```

Further, `Year` is a string, when it should also be an integer. However, this is due to incorrect data entry on some columns. We change `Year` to an integer type and investigate further.

```python
year_numeric = pd.to_numeric(df["Year"], errors="coerce")
not_year_numeric = year_numeric[year_numeric.isna()]
not_year_numeric.shape
```

There are 48k records with NaN entries into `Year`.

```python
df.iloc[not_year_numeric.index]["Year"].unique()
```

These records clearly contain misinputted information into the `Year` column. Since these only represent a small portion of the dataset, we choose to remove them from the cleaned dataset. This is because we assume that `Year` will be some indicator of pricing, which is what we are attempting to predict.

```python
df["Year"] = year_numeric
df = df.iloc[year_numeric.index]
df.head()
```

`NumOwner` still contains N/A entries. However, we will keep the column as is for now. In exploratory analysis, we will identify if there are is a pattern amongst listings with
N/A owners.

### 2.1.3. Is `CarTitle` redundant?

Some of the columns in the dataset are potentially redundant. Namely `CarSpecs`, which is simply a retelling of the data in the dataset in a human-readable format. However, we suspect that `CarTitle` and `CarSubTitle` may also be redudnant too.

We suspect that `CarTitle` is simply a concatenation of the `Make`, `Model`, `Variant` columns.

```python
def normalise(s):
    return (
        s.astype(str)
        .str.lower()
        .str.replace(r"\s+", " ", regex=True)
        .str.strip()
    )

expected_title = normalise(
    df["Make"].astype(str) + " " + df["Model"].astype(str) + " " + df["Variant"].astype(str).fillna("")
)
actual = normalise(df["CarTitle"])

matches = expected_title == actual
matches.mean()
```

```python
df[~matches]
```

So, we find that `CarTitle` is not simple a concatenation of `Make`, `Model` and `Variant`. Notably, there are cases in which the variant is not mentioned in the title. In investigating this, we also discovered that the model column is incomplete. We believe that we are able to use the `CarTitle` to recover this. The `CarTitle` is at least a concatenation of `Make` and `Model`. Therefore, if two records have the same title, then they have the same make and model.

However, we can also conclude that `CarTitle` is most likely data related to the listing, rather than data related to the car itself. This is a useful conclusion to reach when it comes to feature engineering. However, at this stage, we will use it to recover some data.

```python
missing_model = df[pd.isna(df["Model"])]
missing_model
```

We use `CarTitle` to fix the missing model issue as follows:

1. For each `CarTitle` in the dataframe above, check if there a record which has the same `CarTitle` and also includes a `Model`
2. If so, then replace NaN with that model
3. Check what is remaining afterwards, and reassess

```python
# Get a list of title -> models
title_model = df.groupby("CarTitle")["Model"].first()
title_model
```

```python
# Replace NaN models using this data
df.loc[df["Model"].isna(), "Model"] = (
    df.loc[df["Model"].isna(), "CarTitle"].map(title_model)
)

df.head()
```

```python
# Check what's still NaN
missing_model = df[pd.isna(df["Model"])]
missing_model
```

Almost all of the `Model` data was recovered using existing records and the `CarTitle` variable. Only two rows remain with NaN `Model`'s, and the `Model` can easily be identified by eye

- Alpine A110 Legende GT clearly has model "A110", since "Legende GT" is the value of its `Variant`.
- Sunbeam Alpine clearly has model "Alpine" since "Sunbeam" is the `Make`, and there is no `Variant`.

So we can fill these in by hand.

```python
# Manually replace Models for the rows
df.loc[5172, "Model"] = "A110"
df.loc[654124, "Model"] = "Alpine"
```

```python
# Check what's still NaN
missing_model = df[pd.isna(df["Model"])]
missing_model
```

We have managed to recover all `Model` data using the `CarTitle` column. We check if we need to do the same for `Make`.

```python
df[df["Make"].isna()]
```

The `Make` is present for all records.

We shall not do the same for `Variant`. This is because we do not see `Variant` as a reliable variable in the dataset for the following reasons:

- It is frequently missing
- `CarTitle` is associated with the listing and so wording isn't standardised. Thus, it's hard to do it in a way which is automatic
- Manual methods are time consuming and subjective
- There are even some cases where the `Variant` is not mentioned in the `CarTitle`, so even a perfect method could not recover all cases

However, we will still clean the column, since there are instances of an "Unlisted" string which we will replace with NaN now.

```python
df[df["Variant"] == "Unlisted"].shape
```

```python
df.loc[df["Variant"] == "Unlisted", "Variant"] = pd.NA
df[df["Variant"] == "Unlisted"].shape
```

### 2.1.4. Replacing Incomplete Data with NaN

We notice that the `CarBadges` column contains empty strings. We replace these with N/A. We also remove the quotation marks '' where necessary.

```python
# Remove quotes
df["CarBadges"] = df["CarBadges"].str.replace("'", "")
df.head()
```

```python
# Replace empty strings with NaN
df.loc[df["CarBadges"].str.strip().str.len() == 0, "CarBadges"] = df[df["CarBadges"].str.strip().str.len() == 0].map(lambda x: pd.NA)
df.head()
```

### 2.1.5. Column Standardisation

The `Reg` column is a string which always contains the word 'reg' at the end. This word is unnecessary and is removed. We also ensure that all entries are upper case.

```python
df["Reg"] = df["Reg"].str.replace(" reg", "").str.upper()
df.head(20)[["Make", "Model", "Reg", "Year"]]
```

We also standardise the following columns by ensuring all data is given in title case:

- `CarSellerLocation`
- `Transmission`
- `BodyType` 
- `FuelType`

```python
to_title_case = ["CarSellerLocation", "Transmission", "BodyType", "FuelType"]

for column in to_title_case:
    df[column] = df[column].str.title()

df[to_title_case].head(20)
```

### 2.1.5. Cleaning Summary

In this section we have cleaned the data by performing the following:

- Renaming the columns to fit a preferred naming convention
- Fixed a typo in one of the columns
- Examined `CarTitle` and used it to recover missing `Model` information both automatically and manually
- Tidied the `CarBadges` column
- We standardised some of the other columns, e.g. making sure all strings are in title case in location and transmission columns, ensuring `Reg` is always capitalised, and does not include the 'reg' suffix.

We save this clean dataset as `data/clean_car_adverts.parquet`.

```python
df.to_csv("../data/clean_car_adverts.parquet")
```

## 2.2. Feature Engineering

In this section, we clarify what we are attempting to model and select the columns which we are certain will be used for modelling or wish to analyse further before coming to a decision. In addition, we try to "minimise" the dataset. For example, we will take the `EngineSizeUnit` and `EngineSize` columns and attempt to unify them by converting all `EngineSize` measurements into the same units.

### 2.2.1. Problem Definition

We wish to use the car specifications to attempt to predict the selling price of the car. In simpler terms, we wish to use the data which is used to create the advert, not the advert itself. This means that we do not wish to use columns such as `CarAttentionGrabber` and `CarSeller`.

One column which sits in a bit of a grey area is `CarSellerLocation`, since while it is technically tied to the listing rather than being an intrinsic feature of the car. Geographical location could play into some local supply-demand dynamics. Therefore, we shall leave this column.

This is the complete list of columns dropped due to not being relevant for our particular problem:

- `CarTitle`
- `CarSubtitle`
- `CarSpecs`
- `CarBadges`
- `CarAttentionGrabber`
- `CarSeller`
- `CarSellerRating`
- `FinanceAvailable`
- `Discounted`

```python
df.drop(columns=[
    "CarTitle",
    "CarSubTitle",
    "CarSpecs",
    "CarBadges",
    "CarAttentionGrabber",
    "CarSeller",
    "CarSellerRating",
    "FinanceAvailable",
    "Discounted"
], inplace=True)

df.head()
```

### 2.2.2. Engine Power

There are two metrics of engine power (which the dataset refers to as engine size):

1. Brake Housepower (BHP)
2. Pferdestärke (PS)

The dataset makes use of both. For modelling purposes, it is best that we stick to a single metric. Thus, we use the following relation to convert PS into BHP:

$$
    1 \text{ PS} = 0.98632 \text{ BHP}
$$

```python
df["EngineSizeUnit"].unique()
```

```python
df["EngineBHP"] = np.nan

bhp = df["EngineSizeUnit"].str.lower().eq("bhp")
ps = df["EngineSizeUnit"].str.lower().eq("ps")

df.loc[bhp, "EngineBHP"] = df.loc[bhp, "EngineSize"]
df.loc[ps, "EngineBHP"] = 0.98632 * df.loc[ps, "EngineSize"]

df[["Make", "Model", "Variant", "EngineSize", "EngineSizeUnit", "EngineBHP"]].head(20)
```

For every row which has a recorded `EngineSize` and `EngineSizeUnit`, we have converted to BHP. For rows which have at least one of these missing, we have left `EngineBHP` as NaN. We now drop `EngineSize` and `EngineSizeUnit` since they are redundant. If needed in later analysis, they can easily be recovered using the `clean_car_adverts.csv` we created in the last section.

```python
df.drop(columns=[
    "EngineSize",
    "EngineSizeUnit"
], inplace=True)

df.head()
```

### 2.2.3. Summary

In this section we have excluded columns from the dataset which are not relevant for our prediction model, and we have cut down the table size by a single column by converting all engine powers into a single unit. We save the resulting dataset to `data/engineered_car_adverts.parquet`.

```python
df.to_parquet("../data/engineered_car_adverts.parquet")
```
