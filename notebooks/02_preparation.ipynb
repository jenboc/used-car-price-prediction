# 2. Data Preparation

In this notebook we aim to do two things with the raw data `data/all_car_adverts.csv`.

1. Clean the data by fixing the issues we identified in our initial exploration.

```python
import pandas as pd
import numpy as np
import seaborn as sns
```

## 2.1. Column Renaming

In this section we clean the dataset and fix as many of the issues that we identified during initial analysis as possible.

#### Redundant Index Column
Firstly, we identified `Unnamed: 0` as an index column. This is a redundant column since we have access to indices through pandas' API anyway.

```python
df = df.drop(columns=["Unnamed: 0"])
df.head()
```

#### Column Naming

Due to personal preference, we also change the naming convention of the columns from `snake_case` to `CamelCase`.

```python
renaming_map = {
    x: x.replace("_", " ").title().replace(" ", "")
    for x in df.columns.values
}
df = df.rename(columns=renaming_map)

df.head()
```

We rename `Ulez` to `ULEZ` because it is an acronym and we also rename `FeulType` to `FuelType` to fix the typo.

```python
renaming_map = {
    "Ulez": "ULEZ",
    "FeulType": "FuelType"
}
df = df.rename(columns=renaming_map)

df.head()
```

## 2.2. Data Types

We notice that some of the data types seem incorrect. For example, `NumOwner` is clearly not an integer column - but cars can only have an integer amount of owners.

```python
df.dtypes
```

#### Boolean Columns

Some of the `int64` columns only contain 0s or 1s, and thus can be converted to booleans. We check every `int64` column, and convert it to `bool` iff its min and max values are in $\{0,1\}$.

```python
def check_and_change(col_name):
    col = df[col_name]

    if not pd.api.types.is_integer_dtype(col):
        return False
    
    if col.max() > 1 or col.min() < 0:
        return False
    
    df[col_name] = df[col_name].astype(bool)
    return True


for col_name in df.columns.values:
    if check_and_change(col_name):
        print(f"{col_name} changed from an integer column to a boolean column")

df.head()
```

9 columns were changed by this process:

1. `ULEZ`
2. `FullService`
3. `PartService`
4. `PartWarranty`
5. `FullDealership`
6. `FirstYearRoadTax`
7. `BrandNew`
8. `FinanceAvailable`
9. `Discounted`

#### Converting `NumOwner` to Integer

We convert the column `NumOwner` to integer.

```python
df["NumOwner"] = df["NumOwner"].astype(pd.Int64Dtype())
df[["Make", "Model", "Variant", "NumOwner"]].head(20)
```

#### Converting `Year` to Integer

Currently `Year` is a string column. From initial analysis, we know that this is because some entries into the column are erroneous and are misentries belonging to different columns such as `Miles` or `BodyType`. For now, we simply convert to integer, like `NumOwners` and coerce any non-numeric value into NaN. We will concern ourselves with the erroneous numeric entries later.

```python
df["Year"] = pd.to_numeric(df["Year"], errors="coerce").astype(pd.Int64Dtype())
df[["Make", "Model", "Variant", "Year"]].tail(20)
```

## 2.3. Data Recovery

In the initial analysis, we noted that some of the data in the columns are missing. We attempt to recover as much as this as possible automatically.

#### `Model` Recovery

Some rows are missing their `Model` value. Although initial analysis showed that in most cases `CarTitle` is simply a string concatenation of `Make` and `Model`, we cannot reliably recover the data using the column. Hence we take a different approach. There are a lot of duplicate (`Make`, `Model`) pairs in the dataset, many of these also have the same `CarTitle` value. Hence we do the following.

- Construct a map `CarTitle` $\rightarrow$ `Model` using the data in the database, by grouping by `CarTitle` and taking the modal `Model`.
- Use this to fill in missing `Model` entries where possible.
- Manually inspect what remains.

```python
title_to_model = df.groupby("CarTitle")["Model"].apply(
    lambda s: s.mode().iloc[0] if not s.mode().empty else pd.NA
)
title_to_model
```

```python
mask = df["Model"].isna()
pre_map = len(df[mask])

df.loc[
    mask,
    "Model"
] = df.loc[mask, "CarTitle"].map(title_to_model)

mask = df["Model"].isna()
post_map = len(df[mask])

print(f"#NA models pre-map: {pre_map}")
print(f"#NA models post-map: {post_map}")

df[mask][["Make", "Model", "Variant", "CarTitle", "CarSubTitle"]]
```

Automated recovery was successful in all but 2 cases. However we can reasonably recover these models manually by inspecting the columns by eye.

```python
df.loc[5172, "Model"] = "A110"
df.loc[654124, "Model"] = "Alpine"

df[mask][["Make", "Model", "Variant", "CarTitle", "CarSubTitle"]]
```

```python
print(f"Remaining N/A Model Entries: {df["Model"].isna().sum()}")
```

We have successfully recovered `Model` in all cases using a mixture of both automatic and manual techinques.

#### `Variant` (Non-) Recovery

In theory, `Variant` can also be recovered using the `CarTitle` and `CarSubTitle` columns. However, we choose not to here for three reasons:

1. Where the `Variant` value would appear in these columns is not consistent across records, thus an automated method would take a long time to develop since we would have to ensure we are extracting the correct substring.
2. `CarTitle` and `CarSubTitle` pairings may not have many duplicates in the table, meaning a method similar to the one we used to recover `Model` would be inefficient.
3. Initial analysis identified that many entries in the `Variant` column are erroneous (another reason why a `Model`-like method would be unlikely to work) and so it seems that this column is generally unreliable.

For these reasons, we will _not_ recover missing `Variant` entries nor correct the erroneous ones. This therefore means that, unfortunately, we are unable to use this as a feature in modelling.

## 2.4. Calculating New Columns

#### Calculating `EngineBHP`

We introduce a new column into the dataframe which unifies the `EngineSize` and `EngineSizeUnit` columns by converting all engine powers into BHP.

```python
df["EngineSizeUnit"].unique()
```

There are exactly 2 units used: BHP and PS. They are related as follows: 

$$
    1 \text{ PS} = 0.98632 \text{ BHP}
$$

When calculating the new column `EngineBHP`, we set values to NaN if either `EngineSize` or `EngineSizeUnit` is NaN.

```python
df["EngineBHP"] = np.nan

bhp = df["EngineSizeUnit"].str.lower().eq("bhp")
ps = df["EngineSizeUnit"].str.lower().eq("ps")

df.loc[bhp, "EngineBHP"] = df.loc[bhp, "EngineSize"]
df.loc[ps, "EngineBHP"] = 0.98632 * df.loc[ps, "EngineSize"]

df[["Make", "Model", "Variant", "EngineSize", "EngineSizeUnit", "EngineBHP"]].head(20)
```

### 2.2.1. Problem Definition

We wish to use the car specifications to attempt to predict the selling price of the car. In simpler terms, we wish to use the data which is used to create the advert, not the advert itself. This means that we do not wish to use columns such as `CarAttentionGrabber` and `CarSeller`.

One column which sits in a bit of a grey area is `CarSellerLocation`, since while it is technically tied to the listing rather than being an intrinsic feature of the car. Geographical location could play into some local supply-demand dynamics. Therefore, we shall leave this column.

This is the complete list of columns dropped due to not being relevant for our particular problem:

- `CarTitle`
- `CarSubtitle`
- `CarSpecs`
- `CarBadges`
- `CarAttentionGrabber`
- `CarSeller`
- `CarSellerRating`
- `FinanceAvailable`
- `Discounted`

```python
cleaned_df.drop(columns=[
    "CarTitle",
    "CarSubTitle",
    "CarSpecs",
    "CarBadges",
    "CarAttentionGrabber",
    "CarSeller",
    "CarSellerRating",
    "FinanceAvailable",
    "Discounted"
], inplace=True)

cleaned_df.head()
```
