# 5. Conclusions

In this project, we used a dataset of used car sales listings in order to train a model which predicts a price of a used car. We followed a fairly rigorous methodology and started by investigating the factors associated with car prices and attempted to exploit them in order to produce accurate predictions. Additionally, we trained a wide variety of models including linear regression, random forest, gradient boosting and neural network models.

Cross-validation and testing showed that the Random Forest model provided the strongest predictions overall, achieving the following metrics on the test set:

- MAE: £902
- RMSE: £6,199
- $R^2$: 0.924616

Our interpetation of this model was consistent with our earlier analysis of the dataset which identified `Year` and `EngineBHP` as providing the most predictive information. However, we were somewhat surprised that `Make` and `Model` were only deemed modestly important, and we speculated this could be due to technical similarities between different car models across different brands.

## 5.1. Limitations

We note the following:

- We identify predictive relationships and associations, not causal ones.
- The dataset claimed to be from a particular UK used car website, and thus may not generalise universally.
- The dataset had several data-quality concerns (e.g. listing a body type and mileage in the `Year` column as well as missing or incorrect `Variant` entries) and uncertain provenance; the source of the information cannot be independently verified.
- Some potentially relevant variables aren't available (e.g. optional car features such as hill assist, as well as vehicle condition).
